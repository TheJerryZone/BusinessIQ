```python
import sqlite3
import pandas as pd
```

```python
# Connect to SQLite database
# If the file does not exist, SQLite will create it
conn = sqlite3.connect("../database/retailiq.db")
cursor = conn.cursor()
```

```python
# Create a table to store sales data
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales_data (
    order_id TEXT,
    order_date TEXT,
    item_name TEXT,
    item_type TEXT,
    outlet_size TEXT,
    outlet_location TEXT,
    outlet_type TEXT,
    item_visibility REAL,
    rating REAL,
    quantity INTEGER,
    sales_amount REAL,
    month INTEGER,
    year INTEGER
)
""")
```

```python
# Load cleaned Excel data
df = pd.read_excel("../data/Retail_IQ_Sales_Cleaned.xlsx")
```

```python
# Insert data into SQLite table
df.to_sql("sales_data", conn, if_exists="replace", index=False)
```

```python
# Commit changes and close connection
conn.commit()
conn.close()
```
